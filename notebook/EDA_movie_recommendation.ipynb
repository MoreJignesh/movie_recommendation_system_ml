```python
import ast 

def convert_to_list(obj):
    list  = []
    for i in ast.literal_eval(obj):
        # list = list.append(i['name'])
        list.append(i['name'])
    return list
```

```python
new_main_df = main_movies[['movie_id','title','tags']]
new_main_df.head()
```

```python
new_main_df.loc[:,'tags'] = new_main_df['tags'].apply(lambda x: " ".join(x))
```

```python
new_main_df.loc[:,'tags'] = new_main_df['tags'].apply(lambda x: x.lower())
new_main_df['tags'][0]
```

Text Vectorization

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features=5000,stop_words='english')
```

```python
print(new_main_df['tags'].isnull().sum())
print(new_main_df['tags'][0])
```

```python
cv.fit_transform(new_main_df['tags']).toarray().shape
```

```python
vectors = cv.fit_transform(new_main_df['tags']).toarray()
```

```python
vectors[0:2]
```

```python
for word in cv.get_feature_names_out():
    print(word)
```

```python
# !pip install nltk
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

Steming

```python
def stem(text):
    y = []
    for word in text.split():
        y.append(ps.stem(word))

    return " ".join(y)
```

```python
new_main_df.loc[:,'tags'] = new_main_df['tags'].apply(stem)
```

```python
new_main_df.head()
```

```python
from sklearn.metrics.pairwise import cosine_similarity
similarities = cosine_similarity(vectors)
similarities 
```

```python
def recommend(movies):
    movies_index = new_main_df[new_main_df['title'] == movies].index[0]
    distances = similarities[movies_index]
    movies_list = sorted(list(enumerate(distances)),reverse=True,key=lambda x:x[1])[1:6]
    for i in movies_list:
        print(i)
        # new_main_df.iloc[i[0]].title
 
```

```python
# testing
tup = sorted(list(enumerate(similarities[0])),reverse=True,key=lambda x:x[1])[1:6]

for i in tup:
    # i[0]
    print(new_main_df.iloc[i[0]].title)
```

```python
recommend('Batman Begins') 
```
